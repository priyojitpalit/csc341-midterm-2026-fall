## 1. Load the dataset

**Run the next cell exactly as given. Do not edit it.** It imports the required libraries and creates the fixed DataFrame named `df`. Confirm that the printed shape shows **240 rows and 7 columns** before proceeding.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, accuracy_score, log_loss, confusion_matrix

# This cell creates the fixed dataset used throughout the exam. Run it unchanged.
rng = np.random.default_rng(34126)
n = 240

df = pd.DataFrame({
    "practice_hours": rng.integers(1, 13, size=n),
    "sessions_attended": rng.integers(2, 11, size=n),
    "drafts_submitted": rng.integers(0, 5, size=n),
    "feedback_visits": rng.integers(0, 5, size=n),
    "initial_score": rng.integers(40, 86, size=n)
})

noise = rng.normal(0, 5, size=n)
score = (
    12 + 1.7 * df["practice_hours"]
    + 1.2 * df["sessions_attended"]
    + 4.7 * df["drafts_submitted"]
    + 2.6 * df["feedback_visits"]
    + 0.35 * df["initial_score"] + noise
)
df["project_score"] = np.round(np.clip(score, 0, 100), 1)

z = (
    -5.25 + 0.23 * df["practice_hours"]
    + 0.22 * df["sessions_attended"]
    + 0.65 * df["drafts_submitted"]
    + 0.40 * df["feedback_visits"]
    + 0.025 * (df["initial_score"] - 60)
)
probability = 1 / (1 + np.exp(-z))
df["submitted_on_time"] = rng.binomial(1, probability)

display(df.head())
print("Dataset dimensions:", df.shape)
```
